# Inspect Bronze NYC TLC

Notebook para visualizar a Bronze Delta da NYC TLC 2025.

```python
from v2.config.paths import nyc_tlc_bronze_dir

path = nyc_tlc_bronze_dir(2025)
delta_log = path / "_delta_log"

print(f"Path: {path}")
print(f"Path exists: {path.exists()}")
print(f"Delta log exists: {delta_log.exists()}")

if not delta_log.exists():
    raise FileNotFoundError(
        "Tabela Delta Bronze nao encontrada. Rode: poetry run bronze-nyc-tlc"
    )
```

```python
from v2.config.spark import create_spark

spark = create_spark("NotebookInspectBronzeNYCTLC")
```

```python
df = spark.read.format("delta").load(str(path))
df.printSchema()
```

```python
df.show(10, truncate=False)
```
